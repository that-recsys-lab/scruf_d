```python
import json
import plotly.graph_objects as go

# Initialize storage
steps = []
rec_weights = []
rewards = []
running_avg_rewards = []
rbo_scores = []
rbo_running_avg_scores = []  # ✅ NEW
tail_fairness_scores = []
female_fairness_scores = []

# Read rl_scores.log
log_path = "rl_scores.log"

with open(log_path, "r") as f:
    for line in f:
        if "RL_SCORE_ENTRY" in line:
            try:
                entry_start = line.index("{")
                log_entry = json.loads(line[entry_start:])

                steps.append(log_entry["step"])
                rec_weights.append(log_entry["lambda"])
                rewards.append(log_entry["reward"])
                running_avg_rewards.append(log_entry["running_avg_reward"])
                rbo_scores.append(log_entry["rl_scores"]["rbo"]["score"])
                rbo_running_avg_scores.append(
                    log_entry.get(
                        "rbo_running_avg", log_entry["rl_scores"]["rbo"]["score"]
                    )
                )  # ✅ Fallback if not logged
                tail_fairness_scores.append(
                    log_entry["rl_scores"]["fairness"]["tail"]["score"]
                )
                female_fairness_scores.append(
                    log_entry["rl_scores"]["fairness"]["female"]["score"]
                )

            except Exception as e:
                print(f"Error parsing line: {e}")

# Build dynamic plot
fig = go.Figure()

fig.add_trace(
    go.Scatter(x=steps, y=rec_weights, mode="lines+markers", name="Rec Weight (λ)")
)
# fig.add_trace(go.Scatter(x=steps, y=rewards, mode="lines+markers", name="Reward"))
fig.add_trace(
    go.Scatter(
        x=steps, y=running_avg_rewards, mode="lines+markers", name="Running Avg Reward"
    )
)
# fig.add_trace(go.Scatter(x=steps, y=rbo_scores, mode="lines+markers", name="RBO Score"))
fig.add_trace(
    go.Scatter(
        x=steps,
        y=rbo_running_avg_scores,
        mode="lines+markers",
        name="RBO Running Avg",
        line=dict(dash="dot"),
    )
)  # ✅ NEW
fig.add_trace(
    go.Scatter(
        x=steps, y=tail_fairness_scores, mode="lines+markers", name="Tail Fairness"
    )
)
fig.add_trace(
    go.Scatter(
        x=steps, y=female_fairness_scores, mode="lines+markers", name="Female Fairness"
    )
)

fig.update_layout(
    title="RL Scores Over Time",
    xaxis_title="Simulation Step",
    yaxis_title="Value",
    hovermode="x unified",
    legend_title="Metrics",
    template="plotly_white",
    width=1200,
    height=700,
)

fig.show()
```
